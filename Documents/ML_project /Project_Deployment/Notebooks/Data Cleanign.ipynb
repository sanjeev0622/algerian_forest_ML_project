```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
dataset = pd.read_csv("Algerian_forest_fires_dataset_UPDATE.csv",header=1)
```

```python
dataset.info()
```

## data cleaning 

```python
dataset[dataset.isnull().any(axis=1)]
```

```python
dataset.loc[:122,"Region"]=0
dataset.loc[122:,"Region"]=1
```

```python
dataset.head()
df = dataset
```

```python
df.info()
```

```python
df.head()
```

```python
df[["Region"]]=df[["Region"]].astype(int)
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
# remove the null values 
df=df.dropna().reset_index(drop=True)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
## removed teh 122nd row
df=df.drop(122).reset_index(drop=True)
```

```python
df.iloc[[122]]
```

```python
df.head()
```

 ## Changes Required columns as  integer data types

```python
## removing the spaces in the columns
df.columns=df.columns.str.strip()
df.columns
```

```python
df.info()
```

```python
# Change the required columns as integer datatype 
df[['day', 'month', 'year', 'Temperature', 'RH', 'Ws']]=df[['day', 'month', 'year', 'Temperature', 'RH', 'Ws']].astype(int)
```

```python
df.info()
```

## Changing the other columns to float datatype 

```python
objects = [features for features in df.columns
           if df[features].dtypes=='str']
```

```python
objects
```

```python
for i in objects:
    if i != 'Classes':
        df[i]=df[i].astype(float)
```

```python
df.info()
```

```python
df.describe()
```

```python
df.head()
```

```python
## Let ave the cleaned dataset
df.to_csv('Algerian_forest_fires_cleaned_dataset.csv',index=False)
```

## EDA

```python
df.head()
```

```python
## here i droped the starting 3 columns
df_copy = df.drop(['day','month','year'],axis=1)
```

```python
df_copy.head()
```

```python
df_copy['Classes']=np.where(df_copy['Classes'].str.contains('not fire'),0,1)
```

```python
df_copy.head()
```

```python
df_copy['Classes'].value_counts()
```

```python
## percentage for pie chart
percentage = df_copy['Classes'].value_counts(normalize=True)*100
```

```python
## plotting pie chart 
classlabels=["Fire","Not Fire"]
plt.figure(figsize=(12,7))
plt.pie(percentage,labels=classlabels,autopct='%1.1f%%')
plt.title("pie chart os classes")
plt.show()
```
